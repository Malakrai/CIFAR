# Part II — Understanding the CNN

## 7. Convolution layer

For a convolutional layer:

```python
nn.Conv2d(
    in_channels,
    out_channels,
    kernel_size,
    stride,
    padding
)
```

The output spatial dimension is:

$H_{\text{out}}= \left\lfloor\frac{H_{\text{in}} + 2P - K}{S} \right\rfloor + 1$

and similarly for the width.

where:

- \(K\) = kernel size;
- \(S\) = stride;
- \(P\) = padding.

---
Consider:

```python
nn.Conv2d(
    in_channels=1,
    out_channels=16,
    kernel_size=3,
    stride=1,
    padding=1
)
```

Given an input image of:

```text
1 × 28 × 28
```

answer the following **before running any code**:

1. How many kernels are learned?
2. What is the size of each kernel?
3. What will the output tensor shape be for one image?
4. Why does the spatial size remain 28 × 28?

### Answers

1. The layer learns 16 kernels, because out_channels=16. Each kernel acts as a small filter that looks for a specific pattern in the image.

2. Each kernel has a size of 3 × 3 × 3. The 3 × 3 comes from kernel_size=3, and the depth of 3 comes from in_channels=3, since a color image has three channels.

3. The output shape is 16 × 32 × 32. Each of the 16 kernels slides over the whole image and produces one 32 × 32 feature map, which shows where its pattern appears in the image.

4. Using the formula: H_out = ((32 + 2×1 − 3) / 1) + 1 = 32. Without padding, a 3 × 3 kernel cannot be centered on the border pixels, so the output would shrink to 30 × 30. The padding of 1 adds a border of zeros around the image, which allows the kernel to reach the edges. Combined with a stride of 1, the spatial size is preserved.

````python
conv_test = nn.Conv2d(
    in_channels=3,
    out_channels=16,
    kernel_size=3,
    stride=1,
    padding=1,
)

x = torch.randn(1, 3, 32, 32)
y = conv_test(x)

print("Input shape :", x.shape)
print("Output shape:", y.shape)
````

## 8. Pooling

A max-pooling operation such as:

```python
nn.MaxPool2d(kernel_size=2, stride=2)
```

reduces the spatial dimensions.

For example:

```text
16 × 28 × 28
```

becomes:

```text
16 × 14 × 14
```

````python
pool = nn.MaxPool2d(kernel_size=2, stride=2)
# Pooling keeps only the maximum value of each 2x2 square of the feature maps.
# Why: reduces computations and parameters, keeps the strongest features,( on l'applique sur feature maps et non sur l'image)
x = torch.randn(1, 16, 32, 32)
y = pool(x)

print("Before pooling:", x.shape)
print("After pooling :", y.shape)
````

## 9. Activation functions

A convolution is a linear operation. Without nonlinear activation functions, stacking several linear layers would still behave like one large linear transformation.

We will use:

```python
nn.ReLU()
```

## 10. Design the CNN

Create your own network with the final layer contains **10 outputs**, one for each Fashion-MNIST class.

These outputs are called **logits**.

````python
class CifarCNN(nn.Module):
    def __init__(self):
        super().__init__() 

        # Feature extractor: detects patterns in the image
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),   # 3x32x32 -> 16x32x32
            nn.ReLU(),
            nn.MaxPool2d(2),                              # -> 16x16x16

            nn.Conv2d(16, 32, kernel_size=3, padding=1),  # -> 32x16x16
            nn.ReLU(),
            nn.MaxPool2d(2),                              # -> 32x8x8
        )

        # Classifier: turns the features into 10 class scores
        self.classifier = nn.Sequential(
            nn.Flatten(),                 # 32x8x8 -> 2048 values
            nn.Linear(32 * 8 * 8, 128),   # hidden fully connected layer (128 is a hyperparameter)
            nn.ReLU(),
            nn.Linear(128, 10),           # 10 logits, one per class
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


model = CifarCNN().to(device)
print(model)
````

### Explanation of our choices in the CNN

Our network has two parts. The first part (features) contains two convolutional blocks, each made of a convolution, a ReLU activation and a max pooling layer. The first convolution takes the 3 color channels as input and uses 16 filters to detect simple patterns like edges and color contrasts. The second uses 32 filters to combine these patterns into more complex shapes.

We double the number of filters because the image gets smaller after each pooling, so we keep less spatial detail but learn more types of patterns.

We use 3 × 3 kernels with a padding of 1 so that the convolution keeps the same size, and the pooling divides it by 2 each time (32 → 16 → 8). 
`We will test 2 blocks and see the results`.

The second part (classifier) makes the final decision. 

Flatten turns the 32 × 8 × 8 feature maps into a vector of 2048 values. Then a hidden layer of 128 neurons combines these features.

The value 128 is a hyperparameter that we chose ourselves: we start with 128 and we will check the results on the validation set to see if another value works better.

ReLU is used after each layer because without it the network would only be a linear model. The last layer has 10 outputs, one for each class.

## 13. Loss function

This is a 10-class classification problem.
Justify the choice of loss function and try different loss functions.

````python
criterion = nn.CrossEntropyLoss()                      # baseline
# criterion = nn.CrossEntropyLoss(label_smoothing=0.1) 
# criterion = nn.MultiMarginLoss()                    
````

We use `nn.CrossEntropyLoss` because this is a multi-class classification problem with 10 classes and only one correct class per image. It applies a Softmax to the logits to get probabilities, then penalizes the model when the probability of the correct class is low.
A confident but wrong prediction gives a large loss, which produces strong gradients to correct the weights. It also takes the labels directly as class numbers (0 to 9). 

MSE is not suitable here because it is designed for regression, not for classification.

We also try CrossEntropyLoss(label_smoothing=0.1), which prevents the model from being overconfident, and MultiMarginLoss, an SVM-style loss that only requires the correct class score to be higher than the others by a margin.

## 15. Training loop
Train your CNN and then compute the accuracy using:

$\text{Accuracy}=\frac{\text{number of correct predictions}}{\text{number of samples}}\times 100$

````python
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        # TODO 1: Reset gradients.
        optimizer.zero_grad()

        # TODO 2: Forward pass.
        outputs = model(images)

        # TODO 3: Compute loss.
        loss = criterion(outputs, labels)

        # TODO 4: Backpropagation.
        loss.backward()

        # TODO 5: Update network parameters.
        optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    epoch_loss = running_loss / total
    epoch_accuracy = 100.0 * correct / total

    return epoch_loss, epoch_accuracy
````

## 16. Validation function

During validation:

- we do **not** update model parameters;
- gradients are unnecessary;
- dropout and batch-normalization layers, if present, should behave in evaluation mode.


````python
def evaluate(model, loader, criterion, device):
    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            # TODO: Forward pass.
            outputs = model(images)

            # TODO: Compute loss.
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)

            predictions = outputs.argmax(dim=1)
            correct += (predictions == labels).sum().item()
            total += labels.size(0)

    epoch_loss = running_loss / total
    epoch_accuracy = 100.0 * correct / total

    return epoch_loss, epoch_accuracy
````

## 17. Train the network


````python
epochs = 10   #TODO

history = {"train_loss": [],
    "val_loss": [],
    "train_acc": [],
    "val_acc": [],}

for epoch in range(epochs):
    train_loss, train_acc = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device,
    )

    val_loss, val_acc = evaluate(
        model,
        val_loader,
        criterion,
        device,
    )

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)

    print(
        f"Epoch {epoch + 1:2d}/{epochs} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Train Acc: {train_acc:.2f}% | "
        f"Val Acc: {val_acc:.2f}%"
    )
````

We train for 10 epochs. We will check this choice with the loss curves.

## 20. Test-set performance

The test set should be used only after model development is complete.

### Question

Why would repeatedly checking the test accuracy during model development be a bad practice?

The test set is used to measure how the model performs on new images. If we check it many times during development, we start choosing our model based on the test results, so the model becomes adapted to these specific images. The test accuracy is then too optimistic and does not show the real performance. This is why we make all our choices with the validation set and use the test set only once, at the end.

````python
test_loss, test_accuracy = evaluate(
    model,
    test_loader,
    criterion,
    device,
)

print(f"Test Loss    : {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.2f}%")
````

The model reaches a test accuracy of 66.66%. This is very close to the validation accuracy (67.34%), so the validation set was a good estimate of the real performance on new images. The test result is slightly lower, which is normal since our choices were made using the validation set.

The gap between training and test (about 6%) also shows a moderate overfitting.

## 22. Per-class accuracy

A model may perform well overall but poorly on one particular class.
Calculate the classification accuracy separately for each Fashion-MNIST class.
### Questions
1. Which class is easiest for the CNN?
2. Which class is hardest?
3. Why might visually similar classes be confused?

````python
model.eval()

class_correct = [0 for _ in range(10)]
class_total = [0 for _ in range(10)]

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)
        predictions = outputs.argmax(dim=1)

        for label, prediction in zip(labels, predictions):
            label_index = label.item()
            class_total[label_index] += 1

            if prediction.item() == label_index:
                class_correct[label_index] += 1

for i, name in enumerate(class_names):
    accuracy = 100.0 * class_correct[i] / class_total[i]
    print(f"{name:15s}: {accuracy:6.2f}%")
````

3. Some classes look very similar: cats and dogs are both furry animals with four legs, a similar head and similar colors, and birds can be confused with airplanes (sky background) or with other animals in natural backgrounds. Automobiles and trucks also share wheels, windows and similar shapes. At a low resolution of 32 × 32 pixels, the details that separate these classes are often barely visible.

### Comparison with and without dropout

With dropout, the training accuracy is much lower (61.91% vs 73.77% at epoch 10), because half of the hidden neurons are turned off during training. The overfitting disappears: the gap between training and validation goes from about +7% to about −3.5% (the validation accuracy is even higher than the training accuracy).

However, the validation accuracy is slightly lower (65.37% vs 66.77%) and the validation loss is higher (0.994 vs 0.968). The model with dropout learns more slowly, and after 10 epochs it has not learned enough: it is closer to underfitting.

Dropout doesn't always improve performance. It reduces overfitting, but it also slows down learning. Here, it removed the overfitting completely, but with only 10 epochs, the model did not have enough time to reach the performance of the baseline. Since its validation accuracy is still increasing and there is no overfitting, it would probably need more epochs, or a lower dropout rate, to give better results.

### Important

Change **one main factor at a time**. Otherwise, you will not know which modification caused the performance difference.

````python
class ExperimentCNN(nn.Module):
    def __init__(self, c1=16, c2=32, third_conv=False, activation=nn.ReLU):
        super().__init__()

        layers = [
            nn.Conv2d(3, c1, kernel_size=3, padding=1), activation(), nn.MaxPool2d(2),   # -> c1 x 16 x 16
            nn.Conv2d(c1, c2, kernel_size=3, padding=1), activation(), nn.MaxPool2d(2),  # -> c2 x 8 x 8
        ]
        out_channels = c2

        if third_conv:
            # third conv without pooling, to keep an 8 x 8 size
            layers += [nn.Conv2d(c2, 64, kernel_size=3, padding=1), activation()]        # -> 64 x 8 x 8
            out_channels = 64

        self.features = nn.Sequential(*layers)

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(out_channels * 8 * 8, 128),
            activation(),
            nn.Linear(128, 10),
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x
````

````python
def run_experiment(name, change, model, epochs=10, lr=0.001):
    torch.manual_seed(SEED)   # same starting conditions for every experiment
    model = model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = evaluate(model, val_loader, criterion, device)
        print(f"{name} | Epoch {epoch + 1:2d}/{epochs} | Train Acc: {train_acc:.2f}% | Val Acc: {val_acc:.2f}%")

    test_loss, test_acc = evaluate(model, test_loader, criterion, device)
    return {"Experiment": name, "Parameters changed": change,
            "Train Acc": round(train_acc, 2), "Val Acc": round(val_acc, 2), "Test Acc": round(test_acc, 2)}
````

````python
results = []

# Baseline and dropout: already trained, we reuse their results
_, test_acc_dropout = evaluate(model_dropout, test_loader, criterion, device)
results.append({"Experiment": "Baseline", "Parameters changed": "—",
                "Train Acc": round(history["train_acc"][-1], 2), "Val Acc": round(history["val_acc"][-1], 2),
                "Test Acc": round(test_accuracy, 2)})
results.append({"Experiment": "Experiment 1", "Parameters changed": "Dropout p=0.5",
                "Train Acc": round(history_dropout["train_acc"][-1], 2), "Val Acc": round(history_dropout["val_acc"][-1], 2),
                "Test Acc": round(test_acc_dropout, 2)})

# New experiments (one change at a time)
results.append(run_experiment("Experiment 2", "Conv2: 32 -> 64 filters", ExperimentCNN(c2=64)))
results.append(run_experiment("Experiment 3", "Third conv layer (64 filters)", ExperimentCNN(third_conv=True)))
results.append(run_experiment("Experiment 4", "ReLU -> GELU", ExperimentCNN(activation=nn.GELU)))
````

### Results of the experiments

The best validation accuracy is obtained by adding a **third convolutional layer** (69.39%, +2.62% compared to the baseline), followed by increasing the second layer to 64 filters (68.77%, +2.00%). Since CIFAR-10 is complex, giving the model more capacity helps it learn more complex patterns. However, these bigger models also overfit more: the gap between training and validation grows from about 7% to about 11%.

GELU gives a smaller improvement (+1.13%) and the largest overfitting (training accuracy 83.94%, gap of 16%). Dropout removes overfitting (the validation accuracy is even higher than the training accuracy), but with only 10 epochs it learns too slowly and gives a lower validation accuracy (65.37%).

We choose the best model using the validation accuracy, not the test accuracy, so our best architecture is the one with a third convolutional layer. Combining a deeper network with dropout could reduce overfitting while keeping the higher capacity.
